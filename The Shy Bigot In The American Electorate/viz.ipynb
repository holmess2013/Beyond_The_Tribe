```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python

# Lock in the master styling
plt.style.use('btt_style')

# DATA
timeframes = ['1997\n(White Southerners)', '1997-2017\n(National Aggregate)']
bias_levels = [23, -4]

# Setup the figure
fig, ax = plt.subplots(figsize=(6, 6))

# Manually control x-positions
x_positions = [0, 0.4] 

# Plot the bars
bars = ax.bar(x_positions, bias_levels, width=0.2)

# Lock the labels to our custom x-positions
ax.set_xticks(x_positions)
ax.set_xticklabels(timeframes)

# Buffer the x-axis and give the y-axis breathing room for the negative bar
ax.set_xlim(-0.35, 0.75)
ax.set_ylim(-10, 25)

# Ultimate minimalist styling
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['bottom'].set_visible(False)

# --- THE ZERO BASELINE ---
# Draw a segment specifically at zero to ground the bars
ax.hlines(y=0, xmin=-0.15, xmax=0.55, color='black', linewidth=1.5)

# Add the "0%" label to the left of the line
ax.text(-0.20, 0, "0%", 
        ha='right', va='center', 
        fontsize=14, fontweight='bold', color='black')

# Turn off the y-axis entirely and strip the vertical gridlines
ax.get_yaxis().set_visible(False)
ax.grid(False)

# Format x-axis ticks
ax.tick_params(axis='x', length=0, labelsize=12, pad=10)

# Add direct data labels
for bar in bars:
    height = bar.get_height()
    
    # Dynamic logic: if the bar is negative, push the text underneath it
    y_offset = -20 if height < 0 else 8
    va_align = 'top' if height < 0 else 'bottom'
    
    ax.annotate(f"{height}%", 
                xy=(bar.get_x() + bar.get_width() / 2, height), 
                xytext=(0, y_offset),  
                textcoords="offset points", 
                ha='center', 
                va=va_align,
                fontsize=16, 
                fontweight='bold')

# --- STORYTELLING ANNOTATIONS ---
# The dashed gray connector
ax.hlines(y=23, xmin=0, xmax=0.4, color='#7f8c8d', linestyle='--', linewidth=1.5, zorder=1)

# The downward arrow
ax.annotate('', 
            xy=(0.4, -4), 
            xytext=(0.4, 23),  
            arrowprops=dict(arrowstyle="-|>", color='black', lw=2, mutation_scale=20))

# The point change label
ax.text(0.45, 9.5, '-27 pts', 
        ha='left', va='center', 
        fontsize=16, fontweight='bold', color='black')

# Titles and labels
plt.title('The Evaporation of Public Shame:\nEstimated Hidden Racial Prejudice (Percentage Points)', 
          loc='left', fontsize=16, fontweight='bold', pad=25)

width_inches = fig.get_figwidth()
fig.patch.set_linewidth(width_inches * 0.5)

# Export and Show
fig.savefig('comparing_studies.png', dpi=600)
plt.show()
```

```python


# 1. Load your existing BTT style
plt.style.use('btt_style') 

# 2. Data
columns = ['Region', 'Direct\nPoll', 'List\nExperiment', 'The Shy\nBigots']
data = [
    ['South', '19%', '42%', '+23%'],
    ['Non-South', '11%', '0%', '-11%']
]

# 3. Setup the figure
fig, ax = plt.subplots(figsize=(10, 4)) 
ax.axis('off') 

# 4. Add Title (Reduced pad from 25 to 10 for tighter spacing)
ax.set_title("Percentage Of White People Opposed to a Black \nFamily Moving In Next Door", 
             fontsize=22, fontweight='bold', pad=5)

# 5. Create the table
table = plt.table(
    cellText=data,
    colLabels=columns,
    cellLoc='center',
    loc='center',
    colWidths=[0.18, 0.2, 0.25, 0.25] 
)

# 6. Styling
table.auto_set_font_size(False)
table.set_fontsize(20) 
table.scale(1, 4) 

# 7. Bold the header row
for (row, col), cell in table.get_celld().items():
    if row == 0:
        cell.set_text_props(weight='bold')

# 8. Framing logic
fig.patch.set_linewidth(3)
fig.patch.set_edgecolor('gray')
fig.patch.set_visible(True)
width_inches = fig.get_figwidth()
fig.patch.set_linewidth(width_inches * 0.5)

plt.tight_layout(pad=2.0) 
plt.savefig('kulkinski_summary_table.png', dpi=600)
```

```python


df = pd.read_csv("anes_timeseries_cdf_csv_20260205/anes_timeseries_cdf_csv_20260205.csv")

# Apply your BTT style sheet
plt.style.use('btt_style')

# Force to numeric to drop any weird string parsing issues
df['VCF0113'] = pd.to_numeric(df['VCF0113'], errors='coerce') 
df['VCF0105a'] = pd.to_numeric(df['VCF0105a'], errors='coerce') 
df['VCF0303'] = pd.to_numeric(df['VCF0303'], errors='coerce')

# THE FIX: VCF0113 == 1 is South. VCF0105a == 1 is White non-Hispanic.
white_south = df[
    (df['VCF0113'] == 1) & 
    (df['VCF0105a'] == 1) & 
    (df['VCF0303'].isin([1, 2, 3]))
].copy()

# Calculate Percentages by Year
yearly_totals = white_south.groupby('VCF0004').size()
dem_pct = (white_south[white_south['VCF0303'] == 1].groupby('VCF0004').size() / yearly_totals) * 100
rep_pct = (white_south[white_south['VCF0303'] == 3].groupby('VCF0004').size() / yearly_totals) * 100

plot_df = pd.DataFrame({
    'Democrat': dem_pct,
    'Republican': rep_pct
}).dropna()

# Build the Figure 
fig, ax = plt.subplots()

ax.plot(plot_df.index, plot_df['Democrat'], color='#2E74C0', linewidth=3.5, label='Democratic')
ax.plot(plot_df.index, plot_df['Republican'], color='#CB454A', linewidth=3.5, label='Republican')

# Contextual Annotations 
# Thicker, more opaque line so it actually stands out
ax.axvline(x=1997, color='gray', linestyle='--', alpha=0.8, linewidth=2.5)

# Moved up above the data (y=82) and cranked the font size 
ax.text(1998.5, 75, 'Kuklinski Study\n1997', 
        color='#555555', weight='bold', va='center', fontsize=30)
# Formatting 
# Dynamically grab the first and last year in the filtered dataset
min_year = plot_df.index.min()
max_year = plot_df.index.max()

# Inject them into the title using an f-string
ax.set_title(f'White Southern Party Identification ({min_year}–{max_year})', pad=12, loc='left')
ax.set_ylabel('Electorate Share (%)')

# bbox_to_anchor uses (X, Y) coordinates from 0 to 1 relative to the plot area.
# Setting X to 0.98 pushes it to the far right edge, pulling it away from the dotted line.
# Setting Y to 0.05 lifts it just slightly off the floor.
ax.legend(loc='lower right', frameon=False, bbox_to_anchor=(1.02, 0.02))
ax.set_ylim(0, 90)
ax.set_xlim(1946, 2026)

width_inches = fig.get_figwidth()
fig.patch.set_linewidth(width_inches * 0.5)
fig.savefig("white_southerners_leaving.png", dpi=600)
```
