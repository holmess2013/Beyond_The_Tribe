```python
import matplotlib.pyplot as plt
import pandas as pd
from matplotlib.lines import Line2D

# 1. APPLY BTT STYLING
plt.style.use(['Solarize_Light2', 'btt_style'])

# 2. DATA
data = {
    'Year': [1980, 1984, 1988, 1992, 1996, 2000, 2004, 2008, 2012, 2016, 2020],
    'Margin_Error': [5.9, 4.3, 2.8, 2.8, 3.8, 2.9, 2.1, 1.8, 2.9, 2.2, 4.5],
    'Pop_Vote_Correct': [True, True, True, True, True, False, True, True, True, True, True],
    'Predicted_Winner': ['Reagan', 'Reagan', 'Bush', 'Clinton', 'Clinton', 'Bush', 'Bush', 'Obama', 'Obama', 'Clinton', 'Biden']
}
df = pd.DataFrame(data)
historical_avg = df['Margin_Error'].mean()

# Map colors: Green for a correct prediction, Red for a miss
colors = ['#2ecc71' if correct else '#e74c3c' for correct in df['Pop_Vote_Correct']]

# 3. BUILD THE VISUAL
fig, ax = plt.subplots()

# Draw the main line first in a neutral slate color
ax.plot(df['Year'], df['Margin_Error'], linewidth=4, color='#2c3e50', zorder=2)

# Scatter the dots over the line using our new color mapping
ax.scatter(df['Year'], df['Margin_Error'], c=colors, s=500, edgecolors='white', linewidths=2.5, zorder=3)

# --- NEW: DYNAMIC LABELS WITH MANUAL OVERRIDES ---
for i, row in df.iterrows():
    # Set default X position
    x_pos = row['Year']
    
    # If the dot is above or on the historical average line, push text up
    if row['Margin_Error'] >= historical_avg:
        y_pos = row['Margin_Error'] + 0.25
        v_align = 'bottom'
    # If the dot is below the line, pull text down
    else:
        y_pos = row['Margin_Error'] - 0.25
        v_align = 'top'
        
    # --- MANUAL TWEAKS ---
    # Nudge 1980 Reagan to the right
    if row['Year'] == 1984:
        x_pos += 1.2
        
    # Nudge 2012 Obama further down to escape the trendline
    if row['Year'] == 2012:
        y_pos -= 0.25 
        
    ax.text(x_pos, y_pos, row['Predicted_Winner'], 
            ha='center', va=v_align, fontsize=22, color='#2c3e50', zorder=5)
# Average baseline
ax.axhline(historical_avg, linestyle='--', linewidth=3, color='#7f8c8d', zorder=1)
ax.text(2000, historical_avg + 0.4, f'Historical Average ({historical_avg:.1f} pts)', color='#34495e', fontsize=30)

# Axis formatting
ax.set_xticks(df['Year'])
ax.set_xticklabels(df['Year'], rotation=45)
ax.set_xlim(1978, 2022) 
ax.set_ylim(0, 7.5)
ax.set_ylabel('Absolute Margin Error \n(Percentage Points)', labelpad=15)

plt.title("National Polling Accuracy: Error on \nthe Margin (1980 - 2020)", pad=20, loc='left')

# Force the thick outer border
fig.patch.set_edgecolor('black')
fig.patch.set_linewidth(3)

# --- CUSTOM LEGEND ---
legend_elements = [
    Line2D([0], [0], marker='o', color='w', markerfacecolor='#2ecc71', markersize=20, label='Correct Pop. Vote Winner'),
    Line2D([0], [0], marker='o', color='w', markerfacecolor='#e74c3c', markersize=20, label='Wrong Pop. Vote Winner')
]
ax.legend(handles=legend_elements, loc='upper right', fontsize=26, framealpha=0.9, edgecolor='black', borderpad=1)

# THE POWERPOINT FIX
fig.subplots_adjust(left=0.15, right=0.95, bottom=0.15, top=0.80)

# The ultimate anti-cropper save command
fig.savefig('aapor_margin_error_final.png', dpi=300, facecolor='#FDF6E3', edgecolor='black', bbox_inches=None)
plt.show()
```
