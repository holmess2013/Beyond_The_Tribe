```python
import matplotlib.pyplot as plt
import numpy as np

# Drop in your custom BTT style
plt.style.use(['Solarize_Light2', 'btt_style'])

groups = ['Control Group\n(Gift Cards)', 'Treatment Group\n($500/mo)']
start_pct = [32, 28]
end_pct = [37, 40]

y = np.arange(len(groups))
height = 0.35

fig, ax = plt.subplots(figsize=(10, 6), dpi=150)

# Plotting
bars1 = ax.barh(y - height/2, start_pct, height, label='Start of Program', color='#264653')
bars2 = ax.barh(y + height/2, end_pct, height, label='1 Year Later', color='#2a9d8f')

# Formatting labels
ax.set_yticks(y)
ax.set_yticklabels(groups, fontsize=15, fontweight='bold')

# FIX: Force the x-axis to go up to 48 so the 40% label has plenty of margin
ax.set_xlim(0, 48)

# Remove borders
for spine in ax.spines.values():
    spine.set_visible(False)

# Hide the bottom x-axis
ax.xaxis.set_visible(False)

# Data labels
def add_labels(bars):
    for bar in bars:
        width = bar.get_width()
        # FIX: Pushed the text a tiny bit further from the end of the bar (0.8 instead of 0.5)
        ax.text(width + 0.8, bar.get_y() + bar.get_height()/2,
                f'{int(width)}%',
                va='center', ha='left', fontsize=16, fontweight='bold')

add_labels(bars1)
add_labels(bars2)

# Titles
plt.text(-0.2, 1.1, 'SEED: Employment Pre- and Post-Study',
         fontsize=20, fontweight='bold', transform=ax.transAxes)


# Legend
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.05), ncol=2, frameon=False, fontsize=14)

# Invert y-axis
ax.invert_yaxis()

plt.tight_layout()
plt.savefig('stockton_employment_btt.png', bbox_inches='tight')
```
